## PowerSystemResource

All elements of the power system model are categorized as a `PowerSystemResource`, which can be any item of equipment (e.g. a switch), an equipment container with multiple pieces of equipment (e.g. a substation or a feeder), or an organizational entity (e.g. a control area). All power systems modeling components are then divided into two categories, depending on whether the particular focus is on the electrical connectivity or the physical assets.  All electrical aspects of equipment are associated with the classes inheriting from `Equipment` class in the Grid (formerly 61970) package.  

The physical assets associated with a particular piece of equipment are described by the `Asset` class and its children, such as `DuctBank` and `Cabinet`.

The physical nameplate data(e.g. wire diameter) of the equipment are associated with the child classes `AssetInfo` from the Enterprise (formerly 61968) package.

```python
diagram_text = utils.get_mermaid([
    cim.Asset, # The 
    cim.AssetInfo,
    cim.PowerSystemResource,
    cim.Equipment
], show_attributes=False)
Mermaid(diagram_text)
```

## Equipment and ConductingEquipment

Almost all equipment used within a power system network model (such as lines, transformers, loads, generators, and inverters) inherit from the `ConductingEquipment`class. The classes mostly come from the Wires and Generation packages of the Grid (formerly 61970) package. Each of the specific classes are described in more detail in later sections of this tutorial.

```python
diagram_text = utils.get_mermaid([
    cim.ConductingEquipment, # Anything that conducts electricity
    cim.Switch, # All types of switches
    cim.Conductor, # Abstract conductor
    cim.ACLineSegment, # Overhead lines & underground cables
    cim.PowerTransformer, # Transformers
    cim.EnergyConnection, # Abstract class for shunt elements
    cim.EnergyConsumer, # Loads
    cim.RegulatingCondEq, # Abstract class
    cim.RotatingMachine, # Generators and motors
    cim.PowerElectronicsConnection, # Inverters
    cim.ShuntCompensator, # Capacitors & reactors
    cim.EnergySource # Slack bus / infinite source
    ], show_attributes=False)
Mermaid(diagram_text)
```

## Equipment Containers

For the purposes of naming, asset management, and network modeling, all nodes and equipment are associated with a container class. Each node will have a container as specified by the `ConnectivityNode.ConnectivityNodeContainer` association, while equipment will have a primary container specified by `Equipment.EquipmentContainer`.

The relationships between nodes, equipment, and container classes are illustrated below:

```python
diagram_text = utils.get_mermaid([
    cim.ConnectivityNodeContainer, # Top-level container class
    cim.ConnectivityNode, # Electrical bus / node
    cim.Equipment, # Any equipment
    cim.EquipmentContainer # Container of equipment
], show_attributes=False)
Mermaid(diagram_text)
```

### Substation and Feeders

The `Substation` and `Feeder` classes are used to describe containment and naming of equipment within substations and distribution feeders. For most applications, each feeder is described as having a single normal energizing substation that serves as the source for all downstream customers. 

```python
diagram_text = utils.get_mermaid_path(cim.Substation,'NormalEnergizedFeeder')
diagram_text = utils.add_mermaid_path(cim.Feeder, 'NormalEnergizingSubstation', diagram_text)
Mermaid(diagram_text)
```

### Substation Bays and Voltage Levels

In describing large substations with node-breaker models, it is common to define equipment containment based on the physical bay and voltage level such that all conducting equipment, protection relays, measurements, and telemetry are grouped together. The associations between the `Substation`, `Bay`, and `VoltageLevel` classes are illustrated below. The configuration of a particular bay (ring bus, main-and-transfer, etc.) can be specified using the `BreakerConfiguration` and `BusBarConfiguration` enumerations.

```python
diagram_text = utils.get_mermaid([
    cim.Substation,
    cim.VoltageLevel,
    cim.Bay
], show_attributes=True)
Mermaid(diagram_text)
```

It is possible to associate equipment with multiple containers using the `Equipment.AdditionalEquipmentContainer` association. For example, a breaker that is contained within a particular substation bay would be expressed as:

```python
breaker = cim.Breaker(name = 'new_brk',
                      EquipmentContainer=cim.Bay(name = 'bay_1'))
breaker.AdditionalEquipmentContainer.append(cim.Substation(name='north_sub'))
breaker.pprint()
```

It is possible to visualize the relationship to each EquipmentContainer as shown below:

```python
diagram_text = utils.get_mermaid_path(breaker, 'EquipmentContainer')
diagram_text = utils.add_mermaid_path(breaker, ['AdditionalEquipmentContainer','[0]'], diagram_text)
Mermaid(diagram_text)
```
