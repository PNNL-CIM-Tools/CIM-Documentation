## Equipment Containers

For the purposes of naming, asset management, and network modeling, all nodes and equipment are associated with a container class. Each node will have a container as specified by the `ConnectivityNode.ConnectivityNodeContainer` association, while equipment will have a primary container specified by `Equipment.EquipmentContainer`.

The relationships between nodes, equipment, and container classes are illustrated below:

```python
diagram_text = utils.get_mermaid([
    cim.ConnectivityNodeContainer, # Top-level container class
    cim.ConnectivityNode, # Electrical bus / node
    cim.Equipment, # Any equipment
    cim.EquipmentContainer # Container of equipment
], show_attributes=False)
Mermaid(diagram_text)
```

It is possible to associate equipment with multiple containers using the `Equipment.AdditionalEquipmentContainer` association. For example, a breaker that is contained within a particular substation bay would be expressed as:

```python
breaker = cim.Breaker(EquipmentContainer=cim.Bay())
breaker.AdditionalEquipmentContainer.append(cim.Substation())
breaker.pprint()
```

## PowerSystemResource

```python
root = [
    # cim.EquipmentContainer, # Container of equipment
    cim.Substation,
    cim.VoltageLevel,
    cim.Bay,
    cim.BreakerConfiguration
]
set(map(type, root))
```

All elements of the power system model are categorized as a `PowerSystemResource`, which can be any item of equipment (e.g. a switch), an equipment container with multiple pieces of equipment (e.g. a substation or a feeder), or an organizational entity (e.g. a control area). All power systems modeling components are then divided into two categories, depending on whether the particular focus is on the electrical connectivity or the physical assets.  All electrical aspects of equipment are associated with the classes inheriting from `Equipment` class in the Grid (formerly 61970) package.  The physical aspects of equipment are associated with the child classes of `Asset` and `AssetInfo` from the Enterprise (formerly 61968) package.

```python
diagram_text = utils.get_mermaid([
    # cim.IdentifiedObject,
    cim.Asset,
    cim.AssetInfo,
    cim.PowerSystemResource,
    cim.Equipment,
    cim.ConductingEquipment
], show_attributes=False)
Mermaid(diagram_text)
```
